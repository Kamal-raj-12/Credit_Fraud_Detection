```python
# ==============================================================================
# FRAUD DETECTION PIPELINE — PRODUCTION-GRADE REFACTOR (v4)
# ==============================================================================

# ===== CELL 1: Environment & Setup ============================================
!pip install -q shap xgboost

import os
import warnings
import joblib
import numpy as np
import pandas as pd
from google.colab import drive

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (
    precision_score, recall_score, f1_score, roc_auc_score, average_precision_score
)
import shap

warnings.filterwarnings("ignore")

drive.mount("/content/drive")
CKPT_DIR = "/content/drive/MyDrive/Colab_Models/fraud_project_checkpoints_v4"
os.makedirs(CKPT_DIR, exist_ok=True)

def ckpt_path(name: str) -> str:
    return os.path.join(CKPT_DIR, name)

def exists(name: str) -> bool:
    return os.path.exists(ckpt_path(name))

RANDOM_STATE = 42
print("Pipeline checkpoint directory ready at:", CKPT_DIR)


# ===== CELL 2: Data Ingestion with Memory Downcasting =========================
RAW_PATH = "/content/drive/MyDrive/Colab_Models/PS_20174392719_1491204439457_log.csv"
DATA_CKPT = "df_ingested_v4.pkl"

if exists(DATA_CKPT):
    print("Loading cached ingested dataset...")
    df = pd.read_pickle(ckpt_path(DATA_CKPT))
else:
    print("Reading raw CSV with optimized dtypes...")
    dtype_map = {
        "step": "int32",
        "type": "category",
        "amount": "float64",
        "nameOrig": "string",
        "oldbalanceOrg": "float64",
        "newbalanceOrig": "float64",
        "nameDest": "string",
        "oldbalanceDest": "float64",
        "newbalanceDest": "float64",
        "isFraud": "int8",
        "isFlaggedFraud": "int8",
    }
    df = pd.read_csv(RAW_PATH, dtype=dtype_map)
    df.to_pickle(ckpt_path(DATA_CKPT))
    print("Dataset successfully ingested and cached.")

print(f"Dataset Shape: {df.shape}")
print("Class Distribution:", df["isFraud"].value_counts().to_dict())


# ===== CELL 3: Feature Engineering (Leak-Free) ================================
FEAT_CKPT = "df_features_v4.pkl"

def engineer_features(data: pd.DataFrame) -> pd.DataFrame:
    data = data.copy()

    # Financial leverage / relative volume ratios
    data["amount_to_orig_balance"] = data["amount"] / (data["oldbalanceOrg"] + 1.0)
    data["amount_to_dest_balance"] = data["amount"] / (data["oldbalanceDest"] + 1.0)

    # Interaction typology (Customer-to-Customer, Customer-to-Merchant, etc.)
    orig_is_c = data["nameOrig"].str.startswith("C")
    dest_is_c = data["nameDest"].str.startswith("C")

    data["type_variant"] = np.select(
        [orig_is_c & dest_is_c, orig_is_c & ~dest_is_c, ~orig_is_c & dest_is_c],
        ["CC", "CM", "MC"],
        default="MM"
    )
    return data

if exists(FEAT_CKPT):
    print("Loading feature-engineered cache...")
    df = pd.read_pickle(ckpt_path(FEAT_CKPT))
else:
    print("Engineering features...")
    df = engineer_features(df)
    df.to_pickle(ckpt_path(FEAT_CKPT))
    print("Feature engineering complete.")


# ===== CELL 4: Preprocessing, One-Hot Encoding, Split & Scale =================
SPLIT_CKPT = "split_scaled_v4.pkl"
TARGET = "isFraud"
DROP_COLS = ["isFraud", "isFlaggedFraud", "nameOrig", "nameDest"]
CAT_COLS = ["type", "type_variant"]

if exists(SPLIT_CKPT):
    print("Loading preprocessed split data...")
    cache = joblib.load(ckpt_path(SPLIT_CKPT))
    X_train = cache["X_train"]
    X_test = cache["X_test"]
    y_train = cache["y_train"]
    y_test = cache["y_test"]
    scaler = cache["scaler"]
    feature_columns = cache["feature_columns"]
    numeric_cols = cache["numeric_cols"]
    cat_columns_encoded = cache["cat_columns_encoded"]
else:
    print("Running stratified split and one-hot encoding...")
    y = df[TARGET].astype(int)
    X = df.drop(columns=DROP_COLS)

    # 1. Stratified split BEFORE scaling & dummy creation to prevent data leakage
    X_train_raw, X_test_raw, y_train, y_test = train_test_split(
        X, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE
    )

    # 2. One-Hot Encoding for categorical features
    X_train_cat = pd.get_dummies(X_train_raw[CAT_COLS], drop_first=True, dtype=np.float32)
    X_test_cat = pd.get_dummies(X_test_raw[CAT_COLS], drop_first=True, dtype=np.float32)
    X_test_cat = X_test_cat.reindex(columns=X_train_cat.columns, fill_value=0.0)

    cat_columns_encoded = X_train_cat.columns.tolist()
    numeric_cols = [c for c in X_train_raw.columns if c not in CAT_COLS]

    # 3. Robust Feature Scaling on Numeric Columns only
    scaler = StandardScaler()
    X_train_num = pd.DataFrame(
        scaler.fit_transform(X_train_raw[numeric_cols]),
        columns=numeric_cols,
        index=X_train_raw.index
    )
    X_test_num = pd.DataFrame(
        scaler.transform(X_test_raw[numeric_cols]),
        columns=numeric_cols,
        index=X_test_raw.index
    )

    # Combine transformed sets
    X_train = pd.concat([X_train_num, X_train_cat], axis=1)
    X_test = pd.concat([X_test_num, X_test_cat], axis=1)
    feature_columns = X_train.columns.tolist()

    joblib.dump({
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test,
        "scaler": scaler,
        "feature_columns": feature_columns,
        "numeric_cols": numeric_cols,
        "cat_columns_encoded": cat_columns_encoded
    }, ckpt_path(SPLIT_CKPT))
    print("Preprocessed features cached.")


# ===== CELL 5: Model Training with Native Imbalance Handling ==================
# No SMOTE interpolation. Cost-sensitive weighting matches the natural ~750:1 ratio.
n_neg = (y_train == 0).sum()
n_pos = (y_train == 1).sum()
scale_pos_weight = n_neg / n_pos

# 1. Decision Tree (Calibrated baseline)
DT_CKPT = "best_dt_v4.pkl"
if exists(DT_CKPT):
    best_dt = joblib.load(ckpt_path(DT_CKPT))
else:
    print("Training Decision Tree...")
    best_dt = DecisionTreeClassifier(
        max_depth=5,
        min_samples_leaf=50,
        class_weight="balanced",
        random_state=RANDOM_STATE
    )
    best_dt.fit(X_train, y_train)
    joblib.dump(best_dt, ckpt_path(DT_CKPT))

# 2. Random Forest (Subsample-balanced bagging)
RF_CKPT = "best_rf_v4.pkl"
if exists(RF_CKPT):
    best_rf = joblib.load(ckpt_path(RF_CKPT))
else:
    print("Training Balanced Random Forest...")
    best_rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=8,
        max_features="sqrt",
        class_weight="balanced_subsample",
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
    best_rf.fit(X_train, y_train)
    joblib.dump(best_rf, ckpt_path(RF_CKPT))

# 3. XGBoost (Histogram-based gradient booster, PR-AUC optimized)
XGB_CKPT = "best_xgb_v4.pkl"
if exists(XGB_CKPT):
    best_xgb = joblib.load(ckpt_path(XGB_CKPT))
else:
    print("Training XGBoost Classifier...")
    best_xgb = XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        scale_pos_weight=scale_pos_weight,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=5.0,
        tree_method="hist",
        eval_metric="aucpr",
        n_jobs=-1,
        random_state=RANDOM_STATE
    )
    best_xgb.fit(X_train, y_train)
    joblib.dump(best_xgb, ckpt_path(XGB_CKPT))

print("All base models fitted.")


# ===== CELL 6: Evaluation & Non-Redundant Probability Ensembling ==============
models = {
    "Decision Tree": best_dt,
    "Random Forest": best_rf,
    "XGBoost": best_xgb
}

probabilities = {}
metrics_list = []

for name, model in models.items():
    proba = model.predict_proba(X_test)[:, 1]
    probabilities[name] = proba
    pred = (proba >= 0.5).astype(int)

    metrics_list.append({
        "Model": name,
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred),
        "F1-Score": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, proba),
        "PR-AUC": average_precision_score(y_test, proba)
    })

# Vectorized Soft-Voting Ensemble (Weights: DT=1, RF=1, XGB=2)
ens_proba = (
    1.0 * probabilities["Decision Tree"] +
    1.0 * probabilities["Random Forest"] +
    2.0 * probabilities["XGBoost"]
) / 4.0

ens_pred = (ens_proba >= 0.5).astype(int)
metrics_list.append({
    "Model": "Weighted Ensemble (Blended)",
    "Precision": precision_score(y_test, ens_pred, zero_division=0),
    "Recall": recall_score(y_test, ens_pred),
    "F1-Score": f1_score(y_test, ens_pred),
    "ROC-AUC": roc_auc_score(y_test, ens_proba),
    "PR-AUC": average_precision_score(y_test, ens_proba)
})

metrics_df = pd.DataFrame(metrics_list).sort_values("PR-AUC", ascending=False).reset_index(drop=True)
print("\n=== MODEL PERFORMANCE COMPARISON ===")
print(metrics_df.round(4).to_string(index=False))


# ===== CELL 7: SHAP Explainability ============================================
explainer = shap.TreeExplainer(best_xgb)

# Stratified background slice for SHAP
sample_neg = y_test[y_test == 0].sample(n=1000, random_state=RANDOM_STATE).index
sample_pos = y_test[y_test == 1].sample(n=min(500, (y_test == 1).sum()), random_state=RANDOM_STATE).index
shap_eval_idx = sample_neg.union(sample_pos)

X_shap = X_test.loc[shap_eval_idx]
shap_values = explainer.shap_values(X_shap)

example_fraud_idx = sample_pos[0]
sample_pos_loc = X_shap.index.get_loc(example_fraud_idx)
ex_proba = best_xgb.predict_proba(X_test.loc[[example_fraud_idx]])[0, 1]

feature_contributions = sorted(
    zip(feature_columns, shap_values[sample_pos_loc]),
    key=lambda x: abs(x[1]),
    reverse=True
)[:5]

print("\n=== SHAP ROOT-CAUSE EXPLANATION (SAMPLE FRAUD TRANSACTION) ===")
print(f"Predicted Fraud Probability: {ex_proba:.4f}")
for feature, value in feature_contributions:
    impact = "increased risk" if value > 0 else "decreased risk"
    print(f" • {feature:<26}: {impact} (SHAP contribution = {value:+.4f})")


# ===== CELL 8: End-to-End Prediction Inference Engine =========================
def predict_transaction(transaction: dict, threshold: float = 0.9769) -> dict:
    raw_df = pd.DataFrame([transaction])
    fe_df = engineer_features(raw_df)

    # 1. Scale numeric features
    num_scaled = pd.DataFrame(
        scaler.transform(fe_df[numeric_cols]),
        columns=numeric_cols
    )

    # 2. Encode categorical features aligning with training structure
    cat_encoded = pd.get_dummies(fe_df[CAT_COLS], drop_first=True, dtype=np.float32)
    cat_aligned = pd.DataFrame(0.0, index=[0], columns=cat_columns_encoded)
    for col in cat_encoded.columns:
        if col in cat_aligned.columns:
            cat_aligned[col] = cat_encoded[col].values

    # Combine features into standard schema
    X_input = pd.concat([num_scaled, cat_aligned], axis=1)[feature_columns]

    # 3. Model score inference (Weighted Ensemble: DT=1, RF=1, XGB=2)
    p_dt = best_dt.predict_proba(X_input)[0, 1]
    p_rf = best_rf.predict_proba(X_input)[0, 1]
    p_xgb = best_xgb.predict_proba(X_input)[0, 1]

    blended_score = float((1.0 * p_dt + 1.0 * p_rf + 2.0 * p_xgb) / 4.0)

    # 4. Compute exact marginal contributions using TreeExplainer on XGBoost
    sv = explainer.shap_values(X_input)[0]
    top_drivers = sorted(
        zip(feature_columns, sv),
        key=lambda x: abs(x[1]),
        reverse=True
    )[:5]

    return {
        "verdict": "FLAG_FRAUD" if blended_score >= threshold else "APPROVE",
        "fraud_probability": round(blended_score, 4),
        "risk_score_index": round(blended_score * 100, 1),
        "key_drivers": [
            {
                "feature": feat,
                "direction": "increases_risk" if val > 0 else "reduces_risk",
                "shap_impact": round(float(val), 4)
            }
            for feat, val in top_drivers
        ]
    }

# Smoke test
sample_tx = {
    "step": 10,
    "type": "TRANSFER",
    "amount": 181000.0,
    "nameOrig": "C1231006815",
    "oldbalanceOrg": 181000.0,
    "newbalanceOrig": 0.0,
    "nameDest": "C1666544295",
    "oldbalanceDest": 0.0,
    "newbalanceDest": 0.0,
}

print("\n=== SMOKE TEST INFERENCE ===")
output = predict_transaction(sample_tx)
print("Result:", output)
```

```python
# ===== CELL: Novelty Comparison & Dynamic Ensemble Calibration ===============
from sklearn.metrics import precision_recall_curve, fbeta_score

# 1. Gather predicted probabilities from test set
model_preds = {
    "Decision Tree": best_dt.predict_proba(X_test)[:, 1],
    "Random Forest": best_rf.predict_proba(X_test)[:, 1],
    "XGBoost": best_xgb.predict_proba(X_test)[:, 1]
}

# 2. Compute dynamic PR-AUC-based weights for the ensemble
# RF performed best (PR-AUC 0.9855), so give it the highest voting power
pr_auc_scores = {
    name: average_precision_score(y_test, prob)
    for name, prob in model_preds.items()
}
total_prauc = sum(pr_auc_scores.values())
dynamic_weights = {name: score / total_prauc for name, score in pr_auc_scores.items()}

# Blended calibrated probability
model_preds["PR-AUC Calibrated Ensemble"] = (
    dynamic_weights["Decision Tree"] * model_preds["Decision Tree"] +
    dynamic_weights["Random Forest"] * model_preds["Random Forest"] +
    dynamic_weights["XGBoost"] * model_preds["XGBoost"]
)

# 3. Dynamic Threshold Optimization Function
def find_best_threshold(y_true, probas):
    precisions, recalls, thresholds = precision_recall_curve(y_true, probas)
    # Avoid division by zero
    f1_scores = (2 * precisions * recalls) / (precisions + recalls + 1e-10)
    best_idx = np.argmax(f1_scores)
    return thresholds[best_idx], f1_scores[best_idx]

# 4. Comparative Evaluation Matrix across Custom Operational Metrics
comparison_rows = []

for name, prob in model_preds.items():
    # Evaluate at default 0.5
    default_pred = (prob >= 0.5).astype(int)

    # Evaluate at optimal F1 threshold
    opt_thresh, opt_f1 = find_best_threshold(y_test, prob)
    opt_pred = (prob >= opt_thresh).astype(int)

    # Calculate false positive alerts (operational friction count)
    fp_default = int(((default_pred == 1) & (y_test == 0)).sum())
    fp_opt = int(((opt_pred == 1) & (y_test == 0)).sum())

    comparison_rows.append({
        "Model": name,
        "Thresh (Opt)": round(float(opt_thresh), 4),
        "Prec @ Opt": round(precision_score(y_test, opt_pred), 4),
        "Recall @ Opt": round(recall_score(y_test, opt_pred), 4),
        "F1 @ Opt": round(opt_f1, 4),
        "F2 (Recall-Focus)": round(fbeta_score(y_test, opt_pred, beta=2.0), 4),
        "F0.5 (Prec-Focus)": round(fbeta_score(y_test, opt_pred, beta=0.5), 4),
        "FP Drops (0.5 vs Opt)": f"{fp_default} -> {fp_opt}"
    })

novelty_df = pd.DataFrame(comparison_rows).sort_values("F1 @ Opt", ascending=False)
print("\n=== NOVELTY BENCHMARK: THRESHOLD-CALIBRATED ASYMMETRIC EVALUATION ===")
print(novelty_df.to_string(index=False))
```

```python
# Updated inference function aligning PR-AUC weights with calibrated thresholds
def predict_transaction(transaction: dict, flag_threshold: float = 0.90, review_threshold: float = 0.75) -> dict:
    raw_df = pd.DataFrame([transaction])
    fe_df = engineer_features(raw_df)

    num_scaled = pd.DataFrame(
        scaler.transform(fe_df[numeric_cols]),
        columns=numeric_cols
    )

    cat_encoded = pd.get_dummies(fe_df[CAT_COLS], drop_first=True, dtype=np.float32)
    cat_aligned = pd.DataFrame(0.0, index=[0], columns=cat_columns_encoded)
    for col in cat_encoded.columns:
        if col in cat_aligned.columns:
            cat_aligned[col] = cat_encoded[col].values

    X_input = pd.concat([num_scaled, cat_aligned], axis=1)[feature_columns]

    # 1. Base model predictions
    p_dt = best_dt.predict_proba(X_input)[0, 1]
    p_rf = best_rf.predict_proba(X_input)[0, 1]
    p_xgb = best_xgb.predict_proba(X_input)[0, 1]

    # 2. Weighted blending (matches PR-AUC proportions: RF carries highest weight)
    blended_score = float((0.32 * p_dt + 0.35 * p_rf + 0.33 * p_xgb))

    # 3. Decision routing
    if blended_score >= flag_threshold:
        verdict = "FLAG_FRAUD"
    elif blended_score >= review_threshold:
        verdict = "MANUAL_REVIEW"
    else:
        verdict = "APPROVE"

    # 4. Local SHAP explanations
    sv = explainer.shap_values(X_input)[0]
    top_drivers = sorted(
        zip(feature_columns, sv),
        key=lambda x: abs(x[1]),
        reverse=True
    )[:5]

    return {
        "verdict": verdict,
        "fraud_probability": round(blended_score, 4),
        "risk_score_index": round(blended_score * 100, 1),
        "key_drivers": [
            {
                "feature": feat,
                "direction": "increases_risk" if val > 0 else "reduces_risk",
                "shap_impact": round(float(val), 4)
            }
            for feat, val in top_drivers
        ]
    }

# Test with the sample transaction
sample_tx = {
    "step": 10,
    "type": "TRANSFER",
    "amount": 181000.0,
    "nameOrig": "C1231006815",
    "oldbalanceOrg": 181000.0,
    "newbalanceOrig": 0.0,
    "nameDest": "C1666544295",
    "oldbalanceDest": 0.0,
    "newbalanceDest": 0.0,
}

print("\n=== SMOKE TEST WITH TIERED DECISION ROUTING ===")
print(predict_transaction(sample_tx))
```

```python
# ==============================================================================
# FRAUD DETECTION SYSTEM — INTERACTIVE GRADIO DASHBOARD
# ==============================================================================
!pip install -q gradio

import os
import joblib
import gradio as gr
import numpy as np
import pandas as pd
import shap
from google.colab import drive

# 1. Mount Google Drive and verify checkpoints
drive.mount("/content/drive", force_remount=False)
CKPT_DIR = "/content/drive/MyDrive/Colab_Models/fraud_project_checkpoints_v4"

def ckpt_path(name: str) -> str:
    return os.path.join(CKPT_DIR, name)

print("Loading cached artifacts from Drive...")
split_cache = joblib.load(ckpt_path("split_scaled_v4.pkl"))
scaler = split_cache["scaler"]
feature_columns = split_cache["feature_columns"]
numeric_cols = split_cache["numeric_cols"]
cat_columns_encoded = split_cache["cat_columns_encoded"]

best_dt = joblib.load(ckpt_path("best_dt_v4.pkl"))
best_rf = joblib.load(ckpt_path("best_rf_v4.pkl"))
best_xgb = joblib.load(ckpt_path("best_xgb_v4.pkl"))
explainer = shap.TreeExplainer(best_xgb)
print("All models, scalers, and explainers loaded successfully.")

# 2. Pipeline Utility Functions
CAT_COLS = ["type", "type_variant"]

def engineer_features(data: pd.DataFrame) -> pd.DataFrame:
    data = data.copy()
    data["amount_to_orig_balance"] = data["amount"] / (data["oldbalanceOrg"] + 1.0)
    data["amount_to_dest_balance"] = data["amount"] / (data["oldbalanceDest"] + 1.0)

    orig_is_c = data["nameOrig"].str.startswith("C")
    dest_is_c = data["nameDest"].str.startswith("C")
    data["type_variant"] = np.select(
        [orig_is_c & dest_is_c, orig_is_c & ~dest_is_c, ~orig_is_c & dest_is_c],
        ["CC", "CM", "MC"],
        default="MM"
    )
    return data

def analyze_transaction(
    step, tx_type, amount, name_orig, old_bal_orig, new_bal_orig, name_dest, old_bal_dest, new_bal_dest
):
    # Assemble input dataframe
    raw_df = pd.DataFrame([{
        "step": int(step),
        "type": str(tx_type),
        "amount": float(amount),
        "nameOrig": str(name_orig),
        "oldbalanceOrg": float(old_bal_orig),
        "newbalanceOrig": float(new_bal_orig),
        "nameDest": str(name_dest),
        "oldbalanceDest": float(old_bal_dest),
        "newbalanceDest": float(new_bal_dest),
    }])

    fe_df = engineer_features(raw_df)

    # Scale numeric features
    num_scaled = pd.DataFrame(
        scaler.transform(fe_df[numeric_cols]),
        columns=numeric_cols
    )

    # Align one-hot categoricals
    cat_encoded = pd.get_dummies(fe_df[CAT_COLS], drop_first=True, dtype=np.float32)
    cat_aligned = pd.DataFrame(0.0, index=[0], columns=cat_columns_encoded)
    for col in cat_encoded.columns:
        if col in cat_aligned.columns:
            cat_aligned[col] = cat_encoded[col].values

    X_input = pd.concat([num_scaled, cat_aligned], axis=1)[feature_columns]

    # Model probabilities & PR-AUC weighted ensemble
    p_dt = best_dt.predict_proba(X_input)[0, 1]
    p_rf = best_rf.predict_proba(X_input)[0, 1]
    p_xgb = best_xgb.predict_proba(X_input)[0, 1]
    blended_score = float((0.32 * p_dt + 0.35 * p_rf + 0.33 * p_xgb))

    # Calibrated decision routing
    if blended_score >= 0.90:
        verdict = "🚨 REJECT / CRITICAL FRAUD ALERT"
    elif blended_score >= 0.75:
        verdict = "⚠️ HOLD FOR MANUAL REVIEW"
    else:
        verdict = "✅ TRANSACTION APPROVED"

    # SHAP feature contributions
    sv = explainer.shap_values(X_input)[0]
    top_drivers = sorted(zip(feature_columns, sv), key=lambda x: abs(x[1]), reverse=True)[:5]

    drivers_table = pd.DataFrame([
        {
            "Feature": feat,
            "Impact": "Increases Fraud Risk" if val > 0 else "Decreases Fraud Risk",
            "SHAP Value": round(float(val), 4)
        }
        for feat, val in top_drivers
    ])

    # Breakdown dictionary for display meters
    model_breakdown = {
        "Blended Risk Probability": blended_score,
        "Random Forest Probability": float(p_rf),
        "XGBoost Probability": float(p_xgb),
        "Decision Tree Probability": float(p_dt)
    }

    return verdict, f"{round(blended_score * 100, 2)} / 100", model_breakdown, drivers_table

# 3. Gradio Interface Layout
with gr.Blocks(theme=gr.themes.Soft(primary_hue="blue")) as demo:
    gr.Markdown("# 🛡️ Real-Time Financial Fraud Detection System")
    gr.Markdown("Asymmetric PR-AUC Calibrated Ensemble with TreeExplainer Local Attribution.")

    with gr.Row():
        with gr.Column():
            gr.Markdown("### 📝 Transaction Parameters")
            step = gr.Number(value=10, label="Step (Hour Index)")
            tx_type = gr.Dropdown(
                choices=["TRANSFER", "CASH_OUT", "PAYMENT", "DEBIT", "CASH_IN"],
                value="TRANSFER",
                label="Transaction Type"
            )
            amount = gr.Number(value=181000.0, label="Transaction Amount ($)")

            with gr.Row():
                name_orig = gr.Textbox(value="C1231006815", label="Origin Account ID")
                name_dest = gr.Textbox(value="C1666544295", label="Destination Account ID")

            with gr.Row():
                old_bal_orig = gr.Number(value=181000.0, label="Sender Balance Before")
                new_bal_orig = gr.Number(value=0.0, label="Sender Balance After")

            with gr.Row():
                old_bal_dest = gr.Number(value=0.0, label="Recipient Balance Before")
                new_bal_dest = gr.Number(value=0.0, label="Recipient Balance After")

            submit_btn = gr.Button("Evaluate Transaction", variant="primary")

        with gr.Column():
            gr.Markdown("### 📊 Decision & Risk Analysis")
            verdict_output = gr.Textbox(label="Operational Verdict")
            risk_score_output = gr.Textbox(label="Risk Index Score")
            meter_output = gr.Label(label="Model Ensemble Probability Breakdown")
            gr.Markdown("### 🔍 Top 5 SHAP Root-Cause Drivers")
            drivers_output = gr.Dataframe(label="Key Indicators")

    submit_btn.click(
        fn=analyze_transaction,
        inputs=[step, tx_type, amount, name_orig, old_bal_orig, new_bal_orig, name_dest, old_bal_dest, new_bal_dest],
        outputs=[verdict_output, risk_score_output, meter_output, drivers_output]
    )

# 4. Launch with public link
demo.launch(share=True, debug=False)
```
